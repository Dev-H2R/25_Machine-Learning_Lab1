# To implement Find s and candidate elimination algorithm on the enjoy sports dataset and compare the generated hypotheses

```python
X = df.iloc[:, :-1].values
y = df.iloc[:,-1].values
```

```python
import numpy as np
def find_s(X, y):
    hypothesis = None
    print ("initial Hypothesis :",hypothesis)
    for i in range(len(X)):
        if y[i] == "Yes":
            if hypothesis is None:
                hypothesis = X[i]
            else:
                for j in range(len(hypothesis)):
                    if hypothesis[j] != X[i][j]:
                        hypothesis[j] = "?"
        print(f"\nAfter Training Example {i+1}")
        print(hypothesis)
    return hypothesis
final_hypothesis = find_s(X,y)
print("\nFinal Hypothesis :",final_hypothesis)
```

# Candidate Elimination Algorithm

```python
import copy
def candidate_elimination(concepts, target):
    specific = concepts[0].copy()
    general = ["?" for i in range(len(specific))]
    print ("\n Initial S:", specific)
    print ("\n Initial G:", general)
    for i,h in enumerate(concepts):
      if target[i]  == "Yes":
        for x in range(len(specific)):
            if h[x] != specific[x]:
                specific[x] = "?"
        else:
            for x in range(len(specific)):
                if h[x] != specific[x]:
                   general.append(["?" if j!= "?" else specific[x] for j in range(len(specific))])
      print("\nExample", i+1)
      print("S=" , specific)
      print("G=", general)
    return specific,general
S, G = candidate_elimination(X, y)
print("\nFinal Specific Boundary:", S)
print("\nFinal General Boundary:", G)
for g in G:
    print(g)
    for s in S:
        print(s)
```
